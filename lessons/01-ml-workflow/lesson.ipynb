## Machine Learning คืออะไร

Machine Learning หรือ ML คือการให้คอมพิวเตอร์เรียนรู้รูปแบบจากตัวอย่าง เพื่อนำไปทำนายหรือค้นหาโครงสร้างในข้อมูล เช่น ใช้ข้อมูลดอกไม้ที่รู้สายพันธุ์แล้วเพื่อเรียนรู้ความสัมพันธ์ระหว่างขนาดกลีบกับสายพันธุ์ จากนั้นใช้ความสัมพันธ์นั้นทำนายดอกใหม่

**Model** คือกฎหรือความสัมพันธ์ที่ใช้สร้างคำทำนาย ส่วน **algorithm** คือวิธีสร้างโมเดลจากข้อมูล เราต้องตรวจทั้งสิ่งที่เรียนรู้และความสามารถในการใช้กับตัวอย่างใหม่

| ประเภทโจทย์ | คำตอบที่ต้องการ | ตัวอย่าง |
|---|---|---|
| Classification | หมวดหมู่ | สายพันธุ์ดอกไม้ |
| Regression | ค่าปริมาณ | ระยะเวลาจัดส่งเป็นชั่วโมง |
| Clustering | กลุ่มที่ค้นพบจากข้อมูล | กลุ่มพฤติกรรมผู้ใช้โดยไม่ได้ให้คำตอบกลุ่มตอนฝึก |

Classification และ regression ที่เรียนในหลักสูตรเป็น **supervised learning** คือเรียนจากตัวอย่างพร้อมคำตอบ ส่วน clustering เป็นตัวอย่างของ **unsupervised learning** ที่ค้นโครงสร้างโดยไม่ให้คำตอบกลุ่มตอนฝึก กลุ่มที่ค้นพบต้องตีความเพิ่มเติม และไม่จำเป็นต้องตรงกับหมวดหมู่ที่คนกำหนด

## ตั้งโจทย์จากข้อมูล Iris

Iris เป็นชุดข้อมูลที่โหลดได้จาก scikit-learn มีดอกไม้ 150 ดอก แต่ละดอกมีขนาดกลีบ 4 ค่า และทราบสายพันธุ์หนึ่งในสามชนิด: setosa, versicolor หรือ virginica โจทย์บทนี้คือ **ทำนายสายพันธุ์จากขนาดกลีบที่วัดได้**

ก่อนเลือกโมเดล ให้ตอบสี่ข้อ:

| คำถาม | คำตอบสำหรับบทนี้ |
|---|---|
| หนึ่งแถวแทนอะไร | ดอกไม้หนึ่งดอก |
| ต้องการทำนายอะไร | สายพันธุ์ของดอกไม้ |
| ตอนทำนายมีข้อมูลอะไรให้ใช้ | ความยาวและความกว้างของกลีบเลี้ยงและกลีบดอก |
| จะเริ่มวัดผลอย่างไร | สัดส่วนดอกไม้ที่ทายสายพันธุ์ถูก หรือ accuracy |

ในงานจริงต้องพิจารณาด้วยว่าความผิดพลาดแบบใดสำคัญ เช่น ทายผิดแต่ละกลุ่มมีผลต่างกันหรือไม่ เรื่องการเลือกวิธีวัดผลจะขยายในบท 04

## จากตาราง pandas สู่ sample, feature และ target

- **Sample** คือตัวอย่างหนึ่งรายการ ในบทนี้คือดอกไม้หนึ่งดอกหรือหนึ่งแถว
- **Feature** คือข้อมูลที่ใช้ทำนาย ในบทนี้คือขนาดกลีบ 4 คอลัมน์
- **Target** คือคำตอบที่ต้องการทำนาย ในบทนี้คือสายพันธุ์

โค้ดใช้ `load_iris(as_frame=True)` เพื่อรับข้อมูลเป็น pandas ที่คุ้นเคย แล้วแยกข้อมูลกับคำตอบออกจากกัน:

| ชื่อในโค้ด | ความหมาย | ชนิดและรูปร่าง |
|---|---|---|
| `X = data.data` | ตาราง feature | DataFrame ขนาด `(150, 4)` |
| `y = data.target` | คำตอบที่ตรงกับแต่ละแถว | Series ขนาด `(150,)` |
| `data.target_names` | ชื่อสายพันธุ์ตามรหัส | 0 = setosa, 1 = versicolor, 2 = virginica |

`X` และ `y` เป็นชื่อที่นิยมใช้ ไม่ใช่คำสงวนของ Python ส่วน `sklearn` คือชื่อที่ใช้ import ไลบรารี **scikit-learn**

คอลัมน์ `sepal length`, `sepal width`, `petal length`, `petal width` คือความยาว/ความกว้างของกลีบเลี้ยง (sepal) และกลีบดอก (petal) หน่วยเซนติเมตร รหัสสายพันธุ์ 0, 1, 2 เป็นชื่อหมวดหมู่ที่แทนด้วยตัวเลข จึงเป็นโจทย์ classification ไม่ใช่ regression และไม่ควรนำรหัสมาเฉลี่ยเพื่ออธิบายสายพันธุ์

รูปร่าง `(150,)` หมายถึงข้อมูลหนึ่งมิติที่มี 150 ค่า โดยหนึ่งค่าต้องเป็นคำตอบของแถวเดียวกันใน X การมีจำนวนแถวเท่ากันอย่างเดียวยังไม่ยืนยันว่าจับคู่ถูกต้อง ในตัวอย่างนี้ X/y มาจากตารางเดียวกัน และใช้ index เดียวกันตรวจลำดับแถวได้

`assert` ใช้ตรวจเงื่อนไข: หากถูก โปรแกรมทำต่อ หากผิดจะเกิด `AssertionError` ตัวเลข 150 และ 4 ในโค้ดเป็นขนาดเฉพาะของ Iris ส่วนการตรวจ index ช่วยตรวจลำดับแถว แต่ไม่ได้พิสูจน์ว่าคำตอบในข้อมูลต้นฉบับถูกต้องทุกค่า

## แยกข้อมูลฝึกกับข้อมูลทดสอบ

หากฝึกและตรวจด้วยข้อมูลชุดเดิมอย่างเดียว จะยังไม่รู้ว่าโมเดลใช้กับข้อมูลใหม่ได้ดีแค่ไหน จึงกันข้อมูลส่วนหนึ่งไว้ทดสอบ ความสามารถในการนำสิ่งที่เรียนไปใช้กับตัวอย่างใหม่เรียกว่า **generalization**

| ตัวแปร | หน้าที่ |
|---|---|
| `X_train`, `y_train` | ข้อมูลฝึกพร้อมคำตอบ |
| `X_test` | ข้อมูลที่ให้โมเดลลองทำนายหลังฝึก |
| `y_test` | คำตอบจริงที่เก็บไว้ตรวจคำทำนาย |

`train_test_split(X, y, ...)` แบ่ง X กับ y พร้อมกัน เพื่อรักษาคู่ข้อมูลกับคำตอบ และคืนค่าตามลำดับ `X_train, X_test, y_train, y_test`

- `test_size=0.2` กัน 20% ไว้ทดสอบ: train 120 ดอก และ test 30 ดอก สัดส่วนนี้เป็นตัวเลือกของตัวอย่าง ไม่ใช่กฎตายตัว
- `stratify=y` รักษาสัดส่วนคลาสให้ใกล้เคียงข้อมูลเดิม ไม่ได้ทำให้คลาสที่เดิมมีจำนวนน้อยเพิ่มจนเท่าคลาสอื่น
- `random_state=42` ทำให้การสุ่มแบ่งซ้ำได้เมื่อข้อมูลและเงื่อนไขเดิม เลข 42 ไม่ได้ทำให้โมเดลแม่นขึ้น

การแบ่งแบบนี้เหมาะกับตัวอย่างอิสระที่ใช้ฝึกแนวคิดใน Iris ข้อมูลตามเวลาหรือคนเดิมหลายแถวต้องพิจารณาวิธีแบ่งเพิ่มเติมในบท 17

## เริ่มด้วย baseline

**Baseline** คือวิธีง่าย ๆ ที่ใช้สร้างคะแนนอ้างอิง เพื่อดูว่าโมเดลที่ซับซ้อนขึ้นช่วยแก้โจทย์ได้ดีขึ้นหรือไม่ บทนี้ใช้ `DummyClassifier(strategy="most_frequent")` ซึ่งเลือกคลาสที่พบมากที่สุดในชุดฝึก แล้วตอบคลาสนั้นทุกครั้ง โดยไม่ใช้ค่าขนาดกลีบในการตัดสิน

| ขั้น | สิ่งที่เกิดขึ้น |
|---|---|
| `DummyClassifier(...)` | สร้างตัวทำนายและกำหนดวิธีทำงาน |
| `baseline.fit(X_train, y_train)` | เรียนความถี่ของคลาสจากคำตอบในชุดฝึก |
| `baseline.predict(X_test)` | ทำนาย 30 คำตอบ โดยไม่ได้รับ `y_test` |
| `accuracy_score(y_test, pred)` | เปรียบเทียบคำตอบจริงกับคำทำนาย |

`fit` คือขั้นเรียนจากข้อมูล ส่วน `predict` คือขั้นสร้างคำทำนาย ใน scikit-learn จะพบรูปแบบนี้อีกกับโมเดลอื่น

Iris ชุดนี้มี train คลาสละ 40 ดอกเท่ากัน เมื่อเสมอกัน ตัวอย่างนี้เลือกคลาส 0 และทายเป็น setosa ทุกดอก ผลลัพธ์ที่เหมือนกันทุกแถวจึงตรงกับวิธีที่กำหนดไว้ ไม่ใช่การสุ่มทายใหม่ในแต่ละแถว

## วัดผลด้วย accuracy

**Accuracy = จำนวนที่ทายถูก / จำนวนที่ทดสอบทั้งหมด** สำหรับโจทย์จำแนกสายพันธุ์นี้ ค่ายิ่งสูงหมายถึงทายถูกมากขึ้นในชุดที่วัด ค่า 0.333 หรือประมาณ 33.3% เป็นสัดส่วนที่ทายถูก ไม่ใช่ความมั่นใจของโมเดลต่อดอกไม้หนึ่งดอก

ตัวอย่างนี้กำหนด baseline ไว้ก่อนดูผล test เมื่อพัฒนาและเลือกโมเดลจริง ให้ใช้ validation หรือ cross-validation สำหรับปรับวิธี แล้วเก็บ test ไว้ประเมินขั้นสุดท้าย จะได้เรียนรายละเอียดในบท 04 และบท 10

### 1. นำเข้าเครื่องมือ


```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

```

### 2. อ่านข้อมูล Iris และแยก feature กับ target


```python
data = load_iris(as_frame=True)
X = data.data
y = data.target

print(X.head(3).to_string(index=False))
print("First three labels:", y.head(3).to_list())
print("Classes:", dict(enumerate(data.target_names.tolist())))

```

### 3. ตรวจรูปร่างและลำดับแถวของ X/y


```python
print("X shape:", X.shape)
print("y shape:", y.shape)
assert X.shape == (150, 4)
assert y.shape == (150,)
assert X.index.equals(y.index)
print("Shape and index checks passed")

```

### 4. แบ่งข้อมูลพร้อมคำตอบเป็น train และ test


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

```

### 5. ตรวจจำนวนคลาสหลังแบ่งข้อมูล


```python
class_counts = pd.DataFrame({
    "train": y_train.value_counts().sort_index(),
    "test": y_test.value_counts().sort_index(),
})
class_counts.index.name = "class"
print(class_counts.to_string())

```

### 6. ฝึก baseline จากชุด train


```python
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
print("Baseline trained")

```

### 7. ทำนายโดยส่งเฉพาะ X_test


```python
pred = baseline.predict(X_test)
print("Predictions:", pred)
print("Prediction shape:", pred.shape)

```

### 8. เปรียบเทียบคำทำนายกับคำตอบจริง


```python
results = pd.DataFrame({"actual": y_test, "predicted": pred})
results["correct"] = results["actual"] == results["predicted"]
print(results.head(10).to_string())

```

### 9. วัดผลด้วย accuracy


```python
accuracy = accuracy_score(y_test, pred)
print(f"Baseline test accuracy: {accuracy:.3f}")
print(f"Correct percentage: {accuracy:.1%}")

```

### 10. คำนวณ accuracy เองจากจำนวนที่ทายถูก


```python
correct_count = int((pred == y_test).sum())
total_count = len(y_test)
manual_accuracy = correct_count / total_count
print("Correct test predictions:", correct_count, "of", total_count)
print(f"Manual accuracy: {manual_accuracy:.3f}")
print("Same result:", manual_accuracy == accuracy)
```



## อ่านผลที่ได้

| ช่วง | ผลที่คาด | ความหมาย |
|---|---|---|
| โหลดข้อมูล | `X shape: (150, 4)`, `y shape: (150,)` | ดอกไม้ 150 ดอก ขนาดกลีบ 4 ค่าต่อดอก และหนึ่งคำตอบต่อดอก |
| แบ่งข้อมูล | `X_train: (120, 4)`, `X_test: (30, 4)` | กันดอกไม้ 30 ดอกไว้ทดสอบ |
| นับคลาส | train คลาสละ 40, test คลาสละ 10 | สัดส่วนยังเท่ากันเหมือนข้อมูลเดิม |
| ทำนาย | `pred` มี 30 ค่าและเป็น 0 ทุกค่า | baseline ตอบ setosa ทุกดอก |
| วัดผล | `Correct test predictions: 10 of 30` และ accuracy `0.333` | ทายถูกประมาณ 33.3% |
| คำนวณเอง | `Manual accuracy: 0.333` และ `Same result: True` | นับคำตอบที่ตรงกันแล้วหารจำนวนทั้งหมดได้ค่าเดียวกัน |

ตาราง `results` วางคำตอบจริงข้างคำทำนายและ Boolean ที่บอกว่าถูกหรือผิด เลข index ด้านซ้ายเก็บหมายเลขแถวเดิมของข้อมูล test ไว้ การแสดงตารางไม่ได้ส่งเฉลยกลับเข้าโมเดล เพราะทำหลัง `predict` เสร็จแล้ว

`(pred == y_test)` เปรียบเทียบคำตอบทีละตำแหน่ง และ `.sum()` นับ True เป็น 1 จึงนับจำนวนที่ทายถูกได้ ส่วน `:.3f` แสดงทศนิยมสามตำแหน่ง และ `:.1%` แสดงเปอร์เซ็นต์หนึ่งตำแหน่ง

คะแนน baseline ไม่ได้เท่ากับ 33.3% ทุกชุดข้อมูล และการชนะ baseline เพียงอย่างเดียวยังไม่พอจะสรุปว่าพร้อมใช้งานจริง ต้องพิจารณาข้อมูล วิธีวัด และลักษณะความผิดพลาดด้วย

2. หากชุดทดสอบมีคลาส A 90 รายการ และคลาส B 10 รายการ แล้วตัวทำนายตอบ A ทุกครั้ง จะได้ accuracy เท่าไร และยังทำอะไรไม่ได้
3. หากสุ่มลำดับเฉพาะ y แล้วแปลง X กับ y เป็น NumPy array ก่อนจับคู่ตามตำแหน่ง จะเกิดอะไรขึ้น แม้ shape ยังถูกต้อง
4. ตั้งโจทย์ classification หนึ่งอย่าง โดยระบุว่าหนึ่งแถวแทนอะไร ข้อมูลที่มีตอนทำนายคืออะไร คำตอบคืออะไร และจะใช้ baseline ใด พร้อมยกตัวอย่างข้อมูลที่รู้ภายหลังและใช้ทำนายไม่ได้

<details>
<summary>แนวคำตอบ — ลองทำก่อนเปิดอ่าน</summary>

1. X มีความยาว/ความกว้างของกลีบเลี้ยงและกลีบดอก ส่วน y เป็นสายพันธุ์ เลข 0, 1, 2 เป็นรหัสหมวดหมู่ ไม่ใช่ค่าปริมาณที่ต้องการประมาณ
2. Accuracy = 90/100 = 90% แต่ทายคลาส B ผิดทุกครั้ง จึงต้องอ่านคะแนนร่วมกับสัดส่วนคลาสและความผิดพลาดที่สำคัญ
3. โมเดลอาจเรียนจากคำตอบของคนละตัวอย่างโดยไม่เกิด shape error เพราะ array จับคู่ตามตำแหน่ง ควรเลือกหรือแบ่ง X/y พร้อมกัน โดยรักษาแถวให้ตรงกัน การกำหนด Series ให้ DataFrame ใน pandas อาจจับคู่ด้วย index ต่างจาก array; การ reset_index เพียงเพื่อให้เลขตรงกันไม่ได้แก้คู่ข้อมูลที่ผิด
4. เช่น ทำนายว่าพัสดุจะส่งล่าช้าหรือไม่ตอนรับคำสั่งซื้อ หนึ่งแถวคือหนึ่งคำสั่งซื้อ ใช้ระยะทางและบริการที่เลือก คำตอบคือล่าช้า/ไม่ล่าช้า baseline ตอบคลาสที่พบบ่อยที่สุดใน train เริ่มวัด accuracy พร้อมตรวจการพลาดกลุ่มล่าช้า ส่วนเวลาส่งสำเร็จจริงเป็นข้อมูลอนาคต ใช้เป็น feature ณ เวลารับคำสั่งซื้อไม่ได้
